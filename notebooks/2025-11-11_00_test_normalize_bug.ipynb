```python
import os, sys
import trimesh

sys.path.append("..")

from src.utils.data_utils import normalize_mesh, normalize_mesh_robust, mesh_to_surface


# bug_glb_path = "/home/quanta/Downloads/c07706800b4f42689255af5fe311f1e2.glb"
# bug_glb_path = "/home/quanta/Downloads/3071830f2d3e4b57a59976719298c808 (1).glb"


def filter_mesh(obj):
    if isinstance(obj, trimesh.Trimesh):
        return obj.copy()

    if isinstance(obj, trimesh.Scene):
        # 1. 完整复制场景 (geometry + graph)
        #    现在 new_scene 拥有正确的 scale
        new_scene = obj.copy()
        
        # 2. 找出所有 *非* Trimesh 的几何体
        geometries_to_remove = []
        for name, geom in new_scene.geometry.items():
            if not isinstance(geom, trimesh.Trimesh):
                geometries_to_remove.append(name)
        
        # 3. 从复制的场景中删除它们
        #    这会自动清理 graph 中对它们的引用
        if geometries_to_remove:
            new_scene.delete_geometry(geometries_to_remove)
            
        return new_scene
    
    # 如果输入不是 Trimesh 或 Scene，返回一个空场景
    return trimesh.Scene()


bug_glb_path = "/home/quanta/Downloads/3071830f2d3e4b57a59976719298c808.glb"
```

```python
mesh = trimesh.load(bug_glb_path, process=False)
```

```python
mesh_norm_a = mesh.copy()
mesh_norm_a = normalize_mesh(mesh_norm_a)
mesh_norm_a = filter_mesh(mesh_norm_a)
mesh_to_surface(mesh_norm_a.to_geometry(), return_dict=True)['surface_points']
```

```python
mesh_norm_b = mesh.copy()
mesh_norm_b = normalize_mesh_robust(mesh_norm_b)
mesh_norm_b = filter_mesh(mesh_norm_b)
mesh_to_surface(mesh_norm_b.to_geometry(), return_dict=True)['surface_points']
```
